### Fact Sales preparation

- Step 13 — Source 1 ko Join karke Fact Base Banana

first we  orders + customers + order_items + products + sellers + payments ko combine karenge

```python
# ==========================================
# PREPARE FACT SALES - SOURCE 1
# ==========================================

fact_sales_data = order_items.copy()

# ------------------------------------------
# Join Orders
# ------------------------------------------

fact_sales_data = fact_sales_data.merge(
    orders[
        [
            "order_id",
            "customer_id",
            "order_purchase_timestamp"
        ]
    ],
    on="order_id",
    how="left"
)

# ------------------------------------------
# Join Customers
# ------------------------------------------

fact_sales_data = fact_sales_data.merge(
    customers[
        [
            "customer_id"
        ]
    ],
    on="customer_id",
    how="left"
)

# ------------------------------------------
# Join Products
# ------------------------------------------

fact_sales_data = fact_sales_data.merge(
    products[
        [
            "product_id"
        ]
    ],
    on="product_id",
    how="left"
)

# ------------------------------------------
# Join Sellers
# ------------------------------------------

fact_sales_data = fact_sales_data.merge(
    sellers[
        [
            "seller_id"
        ]
    ],
    on="seller_id",
    how="left"
)

# ------------------------------------------
# Join Payments
# ------------------------------------------

fact_sales_data = fact_sales_data.merge(
    payments[
        [
            "order_id",
            "payment_type",
            "installments"
        ]
    ],
    on="order_id",
    how="left"
)

print("Source 1 joins completed!")

print("\nRows:", len(fact_sales_data))
print("Columns:", len(fact_sales_data.columns))

print("\nColumns:")
print(fact_sales_data.columns.tolist())

print("\nFirst 5 rows:")
print(fact_sales_data.head())
```

aur yahan ek important ETL issue mil gaya, jo actually good hai because abhi warehouse mein load karne se pehle fix kar sakte hain.

order_items ke 112,650 rows the, lekin join ke baad 117,604 ho gaye. Reason: ek order_id ke against multiple payment records ho sakte hain. Direct payment join ne kuch order-item rows ko duplicate kar diya.

- Step 14 — Payment duplication fix

Hum payments ko pehle one row per order mein aggregate karenge, phir fact table se join karenge.

```python
# ==========================================
# FIX PAYMENT DUPLICATION
# ==========================================

# One payment record per order
payment_summary = (
    payments
    .groupby("order_id", as_index=False)
    .agg(
        payment_type=("payment_type", "first"),
        installments=("installments", "max")
    )
)

print("Payment summary created!")
print("Payment rows:", len(payment_summary))

# ------------------------------------------
# Rebuild Fact Sales
# ------------------------------------------

fact_sales_data = order_items.copy()

# Join Orders
fact_sales_data = fact_sales_data.merge(
    orders[
        [
            "order_id",
            "customer_id",
            "order_purchase_timestamp"
        ]
    ],
    on="order_id",
    how="left"
)

# Join Payments - now one row per order
fact_sales_data = fact_sales_data.merge(
    payment_summary,
    on="order_id",
    how="left"
)

print("\nFact Sales rebuilt!")
print("Rows:", len(fact_sales_data))
print("Columns:", len(fact_sales_data.columns))

print("\nFirst 5 rows:")
print(fact_sales_data.head())
```

- Step — Foreign Key Integrity Check

```python
# ==========================================
# FOREIGN KEY INTEGRITY CHECK
# ==========================================

fk_check = pd.read_sql(
    """
    SELECT
        COUNT(*) AS total_fact_rows,

        COUNT(*) FILTER (
            WHERE c.customer_key IS NULL
        ) AS missing_customer,

        COUNT(*) FILTER (
            WHERE p.product_key IS NULL
        ) AS missing_product,

        COUNT(*) FILTER (
            WHERE s.seller_key IS NULL
        ) AS missing_seller,

        COUNT(*) FILTER (
            WHERE d.date_key IS NULL
        ) AS missing_date,

        COUNT(*) FILTER (
            WHERE pay.payment_key IS NULL
        ) AS missing_payment,

        COUNT(*) FILTER (
            WHERE m.marketing_key IS NULL
        ) AS missing_marketing,

        COUNT(*) FILTER (
            WHERE del.delivery_key IS NULL
        ) AS missing_delivery,

        COUNT(*) FILTER (
            WHERE r.risk_key IS NULL
        ) AS missing_risk

    FROM fact_sales f

    LEFT JOIN dim_customer c
        ON f.customer_key = c.customer_key

    LEFT JOIN dim_product p
        ON f.product_key = p.product_key

    LEFT JOIN dim_seller s
        ON f.seller_key = s.seller_key

    LEFT JOIN dim_date d
        ON f.date_key = d.date_key

    LEFT JOIN dim_payment pay
        ON f.payment_key = pay.payment_key

    LEFT JOIN dim_marketing m
        ON f.marketing_key = m.marketing_key

    LEFT JOIN dim_delivery del
        ON f.delivery_key = del.delivery_key

    LEFT JOIN dim_risk r
        ON f.risk_key = r.risk_key
    """
    ,
    dw_conn
)

print(fk_check)
```
